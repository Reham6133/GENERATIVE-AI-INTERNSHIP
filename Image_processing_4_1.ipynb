DATA AUGMENTION (editing)

```python
from PIL import Image, ImageEnhance, ImageOps
#PIL = Python imaging library(via pillow) used for image processing
#Image = handles image loading,saving and manipulation
#imageEnhance = provides tools for enhancing image properties like brightness
#ImageOps = contains image processing utilities,such as flipping and grayscaling
#os: a module to handle file system operations like creating directories
```

```python
import os
```

```python
#flips the image horizontally
flipped = ImageOps.mirror(image)
flipped
```

```python
#rotate
rotated = image.rotate(30) #rotates the image by 30 degrees
rotated
```

```python
#brighthness adjustment
enhancer = ImageEnhance.Brightness(image)
#adjusts brightness by a factor of 1.5
brighter = enhancer.enhance(1.5)
brighter
```

```python
#contrast adjustment
contrast = ImageEnhance.Contrast(image).enhance(2.0)
#doubles the image s contrast twice
contrast
```

```python
#grayscale
grayscale = ImageOps.grayscale(image)
grayscale
```
